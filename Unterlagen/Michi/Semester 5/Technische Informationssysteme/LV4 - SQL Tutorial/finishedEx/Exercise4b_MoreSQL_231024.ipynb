## Exercise 1

List all customers by first and last name ordered by name.

```sql
USE AdventureWorksDW2020
GO
SELECT 
    CONCAT(FirstName, ' ', LastName) AS [Name]
FROM DimCustomer 
ORDER BY [Name] ASC
```

## Exercise 2

List only those customers from Canada (CountyRegionCode="CA").

```sql
USE AdventureWorksDW2020
GO
SELECT 
    CONCAT(FirstName, ' ', LastName) AS [Name]
    --,dg.EnglishCountryRegionName AS Country
FROM DimCustomer dc 
INNER JOIN DimGeography dg 
ON dc.GeographyKey = dg.GeographyKey 
AND dg.CountryRegionCode = 'CA'
ORDER BY [Name] ASC
```

## Exercise 3

List the number of customers in each country.



```sql
SELECT 
    COUNT(*) AS [Count],
    dg.EnglishCountryRegionName
FROM DimCustomer dc 
INNER JOIN DimGeography dg 
ON dc.GeographyKey = dg.GeographyKey
GROUP BY dg.EnglishCountryRegionName 
```
