## Data Dictionary

### person

- persno - primary key, personnel number
- deptname - department to which the person belongs
- fname - firstname
- lname - lastname
- salary - salary
- fdate - date at which the person first started at the company
- ldate - date at which the person's employment has been terminated or will be terminated, if null the person is still employed
- gender - gender

### department

- deptno - primary key, department number
- deptname - name of the department
- depthead - head of the department (personnel number)
- budget - monthly budget of the department
- in\_dept - number of the department to which this department belongs (super department)
- founded - founding date of the department

### project

- projno - primary key, project number
- projname - name of the project
- projhead - head of the project (personnel number)
- in\_proj - number of the project to which this project belongs (super project)

### pers\_proj

represents the many to many relationship between projects and persons

- persno - personnel number
- projno - project number
