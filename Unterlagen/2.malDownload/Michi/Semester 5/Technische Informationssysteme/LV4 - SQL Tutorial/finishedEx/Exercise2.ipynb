# 2\. Aggregate Queries

## Task 2.1

Calculate the total from all LineTotal (about all rows) of the SalesOrderDetail table in the Sales schema.

```sql
USE AdventureWorks2017
GO
--This can be achived by adding sum to the following query
SELECT 
    SUM(LineTotal) AS Total
FROM Sales.SalesOrderDetail;

```

## Task 2.2

Calculate the total from all LineTotal (row totals) of the SalesOrderDetail table in the Sales schema.

- Extend the query of task 2.1 to include the SalesOrderID and ProductID columns.
- Now output the list price (sum of LineTotal) for each combination of SalesOrderID and ProductID.
- The column with the sum should be called "Total".

```sql
USE AdventureWorks2017
GO
--This can be achived by adding sum to the following query
SELECT 
    SUM(LineTotal) AS Total,
    SalesOrderID,
    ProductID
FROM Sales.SalesOrderDetail
GROUP BY SalesOrderID, ProductID --alle Felder dei nicht in einer Aggregatfunktion verwendet werden
ORDER BY SalesOrderID;

```

## Task 2.3

Extend the query of Task 2.2 so that only Total greater than $1000 is displayed for each combination of SalesOrderID and ProductID.

```sql
USE AdventureWorks2017
GO
--This can be achived by adding sum to the following query
SELECT 
    SUM(LineTotal) AS Total,
    SalesOrderID,
    ProductID
FROM Sales.SalesOrderDetail
WHERE "LineTotal" >= 1000
GROUP BY SalesOrderID, ProductID
ORDER BY SalesOrderID;


SELECT SalesOrderID, ProductID, SUM(Linetotal) AS 'Total'
FROM Sales.SalesOrderDetail
GROUP BY SalesOrderID, ProductID
HAVING SUM(Linetotal) >= 1000
ORDER BY SalesOrderID;
```

## Task 2.4

- Create a query that displays the minimum, maximum, and average ListPrice for each ProducSubcategoryID from the Product table in the Production schema.
- This should only apply to ProducSubcategoryID that contain a value.

```sql
SELECT  
      ProductSubcategoryID,
      MAX(ListPrice) AS Maximum,
      MIN(ListPrice) AS Minimum,
      AVG(ListPrice) AS Average
  FROM [AdventureWorks2017].[Production].[Product]
  WHERE [ProductSubcategoryID] IS NOT NULL
  GROUP BY ProductSubcategoryID
```

## Task 2.5

- Extend the query from task 2.4 that only results are displayed where the maximum list price is greater than or equal to $1000.
- The results should be displayed in descending order according to the maximum list price.

```sql
SELECT  
      ProductSubcategoryID,
      MAX(ListPrice) AS Maximum,
      MIN(ListPrice) AS Minimum,
      AVG(ListPrice) AS Average
  FROM [AdventureWorks2017].[Production].[Product]
  WHERE [ProductSubcategoryID] IS NOT NULL
  GROUP BY ListPrice, ProductSubcategoryID
  HAVING MAX(ListPrice) >= 1000
  ORDER BY ListPrice DESC;

```
